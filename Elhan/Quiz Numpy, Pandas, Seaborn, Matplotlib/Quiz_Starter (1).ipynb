## Section B — Pandas

### B1. Explore the data

```python
# (a) rows & columns
print(df.shape)
# (b) missing values per column
na_data = df.isnull().sum()
print(na_data)
# (c) game with highest Rating
max_data = df.loc[df['Rating'].idxmax(), 'Rating']
print(max_data)
```

### B2. Filter like a detective

```python
filtered_games = df[(df['Multiplayer'] == 'Yes') & (df['Price'] < 40) & (df['Rating'] >= 8)]
game_count = len(filtered_games)
print(filtered_games)
```

### B3. Group and compare

```python
# (a)
df_agg = df.groupby('Genre')['Copies_Sold_Millions'].mean().sort_values(ascending=False)
print(df_agg)
# (b)
df_2agg = df['Platform'].value_counts()
print(df_2agg)
# (c)
df_3agg = df.groupby('Platform')['Price'].max()
print(df_3agg)
```

### B4. Create new columns

```python
# (a) Revenue_Millions
df['Revenue_Millions'] = df['Price'] * df['Copies_Sold_Millions']
# (b) Age
df['Age'] = 2026 - df['Release_Year']
# (c) Is_Free
df['Is Free'] = df['Price'] == 0
# (d) top 5 by Revenue_Millions
top_5 = df.nlargest(5, 'Revenue_Millions')
print(top_5)
```

### B5. Clean the data

```python
# (a)
df['Rating'] = df['Rating'].fillna(df['Rating'].mean())
# (b)
df['Hours_To_Beat'] = df['Hours_To_Beat'].fillna(df['Hours_To_Beat'].median())
# (c)
df = df[df['Hours_To_Beat'] != 999]
# (d) Why median instead of mean?  ->  We used the median because there was an outlier in the dataframe and if we had used mean, the average would have skewered on one side.
```

## Section C — Matplotlib

### C1. Bar chart

```python
genre_counts = df['Genre'].value_counts()
plt.barh(genre_counts.index, genre_counts.values, color = 'blue')
plt.title('Which Genre has the most games?')
plt.xlabel('Genre')
plt.ylabel('Number of Games')
plt.show()
```

### C2. Scatter plot

```python

# Do cheaper games sell more? What about free games?  ->  The cheaper a game is, the more sales it has. Also, free games have the most sales.

plt.scatter(df['Price'], df['Copies_Sold_Millions'], color = 'blue')
plt.title('Do Cheaper Games sell more?')
plt.xlabel('Price')
plt.ylabel('Copies_Sold_Millions')
plt.show()
```

### C3. Two plots side by side

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df['Rating'], bins = 10, color = 'blue')
axes[0].set_title('Ratings')
axes[0].set_xlabel('Rating')
axes[0].set_ylabel('Count')

year_counts = df['Release_Year'].value_counts().sort_index()
axes[1].plot(year_counts.index, year_counts.values, color = 'blue', marker = 'o')
axes[1].set_title('Games Released per Year')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Number of Games')
```

## Section D — Seaborn

### D1. Count plot

```python

# Platform with most single-player games ->  Mobile
sns.countplot(data=df, x='Platform', hue='Multiplayer')
plt.title('Multiplayer Status')
```

### D2. Box plot

```python

# Highest median rating genre ->  RPG
# Widest spread genre ->  Simulation
sns.boxplot(data=df, x='Genre', y='Rating')
plt.title('Rating of Each Genre')
plt.xticks(rotation=45)
```

### D3. Heatmap

```python

# Most strongly correlated pair -> Hours to beat and Revenue in Millions
# Positive or negative? -> Positive
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
```
