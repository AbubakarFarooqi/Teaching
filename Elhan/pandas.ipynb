```python
import pandas as pd
```

```python
data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
```

```python
print(series[2])
```

change the index

```python
data = [1,2,3,4,5]
index = ['a','b','c','d','e']
series = pd.Series(data,index)
print(series)
```

```python
series['name'] = 'Ali'
```

Dataframes

```python
data={
    'Name':['Azan','Ali','Elhan'],
    'Age':[12,13,14],
    'City':['Lahore','Chennai','kerela']
} # Dictionary
index = ['Azan','Ali','Elhan']
df = pd.DataFrame(data,index=index)

print(df)
```
